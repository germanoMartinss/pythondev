```python
import re
```

```python
def extrair_urls(texto):
    padrao = r"https?://[a-zA-Z0-9.-]+\.[a-zA-Z]{2,}"
    urls = re.findall(padrao, texto)
    return urls
```

```python
texto_exemplo = """
Visite nosso site em https://www.meusite.com.br
e http://blog.exemplo.org para mais informações.
Entre em contato conosco em https://aqui.com
"""

urls_encontradas = extrair_urls(texto_exemplo)
print(urls_encontradas)
```

```python
def validar_senhas(senhas):
    padrao = r"^(?=.*[A-Z])(?=.*[^a-zA-Z0-9]).{8,}$"
    return bool(re.match(padrao, senhas))

senhas = ["Senha@123", "senha@123", "Senha1234", "Se@1", "Abcdefg!"]
for senha in senhas:
    print(f"A senha {senha} é válida? {validar_senhas(senha)}")
```

```python
def contar_palavras(texto):
    padrao_palavra = r"\b\w+\b"
    palavras = re.findall(padrao_palavra, texto)
    return len(palavras)
```

```python
texto_exemplo = """
Ola, tudo bem?
Estou aprendendo Python.
Estou gostando muito!
"""
palavras_encontradas = contar_palavras(texto_exemplo)
print(f"Numero de palavras encontradas: {palavras_encontradas}")
```
