Walrus
- Esse operador é usado para atribuição de variaveis dentro de expressões
- Permite que você atribua um valor a uma variavel enquanto avalia a expressão
- É usado com o operador :=

```python
numeros = [5, 12, 3, 18, 8, 20]
```

```python
# Filtar e armazenar números maiores que 10


numeros_maiores = [n for n in numeros if(maior := n) > 10]
numeros_maiores
```
