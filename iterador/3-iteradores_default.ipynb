```python
import itertools
```

```python
#Exemplo 1 - Range()

for i in range(3):
    print(i)
```

```python
#Exemplo 2 - Iterar Lista

frutas = ["banana", "maca", "laranja", "abacaxi"]
for fruta in frutas:
    print(fruta)
```

```python
#Exemplo 3 - Iterar Dicinário

pessoa = {'nome': 'Alice', 'idade': 25}
for chave in pessoa:
    print(chave, pessoa[chave])
```

```python
#Exemplo 4 - Enumerate()

for i, fruta in enumerate(frutas):
    print(i, fruta)
```

```python
#Exemplo 5 - Zip()

nomes = ["Alice", "Bob", "Charlie"]
idades = [30, 25, 35]
for nome, idade in zip(nomes, idades):
    print(f"{nome} tem {idade} anos.")
```

```python
#Exemplo 6 - Itertools

for i in itertools.count(start=10, step=2):
    if i > 20:
        break
    print(i)
```
