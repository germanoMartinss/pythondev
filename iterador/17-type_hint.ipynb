Type Hint
- É uma maneira de indicar um tipo de variáveis, parâmetros e valores de retorno funções

```python
def soma(a: int, b: int) -> int:
    return a + b

soma(5, 521)
```

```python
from typing import List
```

```python
class Estudante:
    def __init__(self, nome: str, notas: List[float]) -> None:
        self.nome = nome
        self.notas = notas

    def media(self) -> float:
        return sum(self.notas) / len(self.notas)
```

```python
def alunos_acima_da_media(alunos: List[Estudante], media_minima: float) -> List[str]:
    return [aluno.nome for aluno in alunos if aluno.media() >= media_minima]
```

```python
alunos = [
    Estudante("Germano", [9.9, 8.9, 9.0]),
    Estudante("Fernando", [6.6, 10, 9.9]),
    Estudante("João", [7.7, 8.8, 9.9]),
    Estudante("Maria", [10, 5.5, 4.3])
]
```

```python
resultados = alunos_acima_da_media(alunos, 9.5)
resultados
```
