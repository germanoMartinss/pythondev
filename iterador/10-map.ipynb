```python
def quadrado(x):
    return x ** 2
```

```python
numeros = [1, 2, 3, 4, 5]
quadrados = map(quadrado, numeros)
print(list(quadrados))
```

```python
def celsius_to_fahrenheit(celsius):
    return celsius * 9/5 + 32

temperaturas = [0, 10, 20, 30, 40]
fahrenheit = map(celsius_to_fahrenheit, temperaturas)
print(list(fahrenheit))
```

```python
def normalize_name(name):
    return name.strip().lower()

nomes = [" João ", "    Maria ", " Joaquim    ", "GERMANO    ", " JUlianA"]
nomes_normalizados = map(normalize_name, nomes)
print(list(nomes_normalizados))
```
