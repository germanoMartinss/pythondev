Programação funcional
- Trata a computação como a avalição de funções matemáticas e evita estados e dados mutáveis

```python
# Função de alta ordem - é uma função que recebe outra função como argumento

def quadrado(x):
    return x * x

def aplicar_funcao(funcao, lista):
    return [funcao(x) for x in lista]
```

```python
numeros = [1, 2, 3, 4, 5]
resultado = aplicar_funcao(quadrado, numeros)
resultado
```

```python
usuarios = [
    {"nome": "Germano", "idade": 35},
    {"nome": "Fernanda", "idade": 31},
    {"nome": "Mauricio", "idade": 65},
    {"nome": "Maria Julia", "idade": 70}
]
```

```python
def extrair_nomes(usuarios):
    return list(map(lambda u: u['nome'], usuarios))

def filtrar_maiores_que(usuarios, idade):
    return list(filter(lambda u: u['idade'] > idade, usuarios))

def calcular_media_idade(usuarios):
    if not usuarios:
        return 0
    total_idade = sum(map(lambda u: u['idade'], usuarios))
    return total_idade / len(usuarios)
```

```python
usuarios_maiores_que_40 = filtrar_maiores_que(usuarios, 40)
usuarios_maiores_que_40
```

```python
nomes_maiores_que_27 = extrair_nomes(usuarios_maiores_que_40)
nomes_maiores_que_27
```

```python
meida_idade = calcular_media_idade(usuarios_maiores_que_40)
meida_idade
```

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)
```

```python
for i in range(10):
    print(f"Fibonnaci({i}) = {fibonacci(i)}")
```
